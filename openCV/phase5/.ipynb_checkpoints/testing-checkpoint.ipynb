```python
import cv2

import numpy as np
```

```python
image = cv2.imread(r"C:\Users\anubh\OneDrive\Downloads\testing_image.jpg",cv2.IMREAD_GRAYSCALE)
```

```python
if image is None:
    print("❌ Image not loaded. Check path.")
else:
    cv2.imshow("Window Title", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
edges = cv2.Canny(image , 50 , 150)


cv2.imshow("Window Title", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
ret, thresh_img = cv2.threshold(image,120,255,cv2.THRESH_BINARY)

cv2.imshow("Window Title", thresh_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img1 = np.zeros((300,300),dtype="uint8")
img2 = np.zeros((300,300),dtype="uint8")

cv2.circle(img1,(150,150),100 ,255,-1)
cv2.rectangle(img2,(100,100),(250,250) ,255,-1)


cv2.imshow("Window Title", img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow("Window Title", img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Bitwise Operations

bitwise_and = cv2.bitwise_and(img1,img2)
cv2.imshow("Window Title", bitwise_and)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

bitwise_or = cv2.bitwise_or(img1,img2)
cv2.imshow("Window Title", bitwise_or)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
bitwise_not = cv2.bitwise_not(img1)
cv2.imshow("Window Title", bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
