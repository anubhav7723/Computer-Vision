```python
import cv2

import numpy as np
```

```python
image = cv2.imread(r"C:\Users\anubh\OneDrive\Downloads\testing_image.jpg")

if image is None:
    print("❌ Image not loaded. Check path.")
else:
    cv2.imshow("Window Title", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# making binary Image 

binary_img = cv2.Canny(image , 120 , 250)

if binary_img is None:
    print("❌ Image not loaded. Check path.")
else:
    cv2.imshow("Window Title", binary_img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
grey = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
```

```python
#image thresholding

_ , threshold_img = cv2.threshold(grey ,100,255,cv2.THRESH_BINARY)
cv2.imshow("Window Title", threshold_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Finding Contours

contours , hierarchy = cv2.findContours(threshold_img ,cv2.RETR_TREE,cv2.CHAIN_APPROX_SIMPLE)

# draw contours

cv2.drawContours(image , contours , -1 , (0,255,0) , 3)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
for contour in contours:
    approx = cv2.approxPolyDP(contour,0.01*cv2.arcLength(contour,True),True)

    corners = len(approx)

    

    if(corners == 3):
        print("Shape = Triangle")
    
```
