```python
import cv2
```

```python
image = cv2.imread(r"C:\Users\anubh\OneDrive\Downloads\testing_image.jpg")
```

```python
if image is None:
    print("❌ Image not loaded. Check path.")
else:
    cv2.imshow("Window Title", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
image.shape
```

```python
#rectangle by lines
color = (255,255,255)
thickness = 2

cv2.line(image,(50,80),(350,80),color,thickness)
cv2.line(image,(350,80),(350,330),color,thickness)
cv2.line(image,(50,80),(50,330),color,thickness)
cv2.line(image,(50,330),(350,330),color,thickness)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#draw rectangle

cv2.rectangle(image,(350,330),(50,80),color,thickness)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#draw circle

cv2.circle(image,(150,150),100,color,thickness)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# adding text
cv2.putText(image , "Strawberry",(100,100),cv2.FONT_HERSHEY_PLAIN,1.2,color)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# draw all shapes

cv2.rectangle(image,(350,330),(50,80),color,thickness)
cv2.line(image,(50,80),(350,330),color,thickness)
cv2.circle(image,(200,200),120,color,thickness)
cv2.putText(image , "Strawberry",(140,75),cv2.FONT_HERSHEY_PLAIN,1.5,color)

cv2.imshow("Window Title", image)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite("shaped_image.jpg",image)
```
