```python
import cv2
```

```python
image = cv2.imread(r"C:\Users\anubh\OneDrive\Downloads\testing_image.jpg")
```

```python
if image is None:
    print("❌ Image not loaded. Check path.")
else:
    cv2.imshow("Window Title", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
image.shape
```

```python
#(width,height)
resized = cv2.resize(image , (500,300))

cv2.imshow("Window Title", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()


cv2.imwrite("resized_image.jpg",resized)
```

```python
#image[y: y , x:x]
cropped_image = image[80:340,50:350]

cv2.imshow("Window Title", cropped_image)
cv2.waitKey(0)
cv2.destroyAllWindows()


cv2.imwrite("cropped_image.jpg",cropped_image)
```

```python
#image Rotation

h,w = image.shape[:2]
center = (w//2,h//2)

M = cv2.getRotationMatrix2D(center , 90 , 1.0)
rotated_image = cv2.warpAffine(image , M , (w,h))

cv2.imshow("Window Title", rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite("rotated_image.jpg",rotated_image)
```

```python
#image flipped



flipped_horizontal = cv2.flip(image , 1)
flipped_vertical = cv2.flip(image , 0)
flipped_both = cv2.flip(image , -1)

cv2.imshow("Window Title", flipped_both)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite("flipped_horizontal.jpg",flipped_horizontal)
cv2.imwrite("flipped_vertical.jpg",flipped_vertical)
cv2.imwrite("flipped_both.jpg",flipped_both)
```
